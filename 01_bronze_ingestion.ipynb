```python
from pyspark.sql import functions as F

raw_path = "/Volumes/car_sales/bronze/raw_files/vehicles.csv"

df_raw = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "false")
    .option("multiLine", "true")
    .option("quote", '"')
    .option("escape", '"')
    .csv(raw_path)
)

print("Rows:", df_raw.count())
print("Columns:", len(df_raw.columns))

df_raw.printSchema()
```

```python
df_bronze = (
    df_raw
    .withColumn("_ingestion_timestamp", F.current_timestamp())
    .withColumn("_source_file", F.lit("vehicles.csv"))
)

print("Bronze rows:", df_bronze.count())
print("Bronze columns:", len(df_bronze.columns))
```

```python
spark.sql("""
    SELECT COUNT(*) AS row_count
    FROM car_sales.bronze.vehicles_raw
""").show()
```

```python
spark.sql("""
    DESCRIBE TABLE car_sales.bronze.vehicles_raw
""").show(truncate=False)
```

```python
(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("car_sales.bronze.vehicles_raw")
)
```
